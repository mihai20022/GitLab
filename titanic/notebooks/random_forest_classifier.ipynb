```python
import pandas as pd
train_data = pd.read_csv("../data/train.csv") #to understand how paths are working
test_data = pd.read_csv("../data/test.csv") #to understand how paths are working
```

```python
y_train=train_data["Survived"]
features = ["Pclass", "Sex", "SibSp", "Parch"]
X_train = pd.get_dummies(train_data[features])
X_test = pd.get_dummies(test_data[features])

```

```python
X_train
```

```python
from sklearn.ensemble import RandomForestClassifier 


# Trying the Random Forest Classifier
model = RandomForestClassifier(n_estimators=100, random_state=1, max_depth=5)
model.fit(X_train, y_train)
y_test = model.predict(X_test)

output = pd.DataFrame({"PassengerId": test_data.PassengerId, "Survived": y_test})
output.to_csv("../data/submission.csv", index=False)



```
